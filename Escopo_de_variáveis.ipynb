```python
#correção do erro
#primeira forma para a função
def calcular_total(precos):
    total = 0

    for preco in precos:
        total = total + preco

    return total

#segunda forma para a função
def calcula_total(precos):
  global total

  for preco in precos:
        total = total + preco



#primeira forma para o código
total = calcular_total([10, 20, 30])
print(total)

#segunda forma para o código
total = 0
calcula_total([10, 20, 30])
print(total)
```
