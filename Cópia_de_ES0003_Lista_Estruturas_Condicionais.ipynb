## 05. Distância euclidiana e pontos coincidentes

Sabe-se que no cálculo da distância euclidiana entre dois pontos, cada ponto é definido pelas coordenadas `(x, y)`. A distância pode ser calculada de acordo com a fórmula:

$$d(P,Q) = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2}$$

Escreva um programa em Python que informe a distância euclidiana de dois pontos fornecidos pelo usuário. Em seguida, informe se esses pontos são coincidentes.

**Dica:** Os pontos são denominados coincidentes quando sua distância é igual a zero.

```python
# Resposta da questão 05
from math import sqrt
x1 = eval(input("x1: "))
y1 = eval(input("y1: "))
x2 = eval(input("x2: "))
y2 = eval(input("y2: "))
d = sqrt((x1 - x2)**2 + (y1 - y2)**2)
if d == 0:
  print("Os pontos são coincidentes!")
else:
  print("Os pontos não são coincidentes!")
```

## 06. Três pontos formam um triângulo?

Escreva um programa que leia três pontos geométricos quaisquer e determine se estes formam um triângulo. Para isto, pode ser utilizada a fórmula da distância entre dois pontos para calcular as medidas dos lados do triângulo.

**Dica:** Note-se que um triângulo só é válido se a medida de cada um dos seus lados é menor que a soma dos lados restantes.

```python
# Resposta da questão 06
from math import sqrt
x1 = eval(input("x1: "))
y1 = eval(input("y1: "))
x2 = eval(input("x2: "))
y2 = eval(input("y2: "))
x3 = eval(input("x3: "))
y3 = eval(input("y3: "))
d12 = sqrt((x1 - x2)**2 + (y1 - y2)**2)
d23 = sqrt((x2 - x3)**2 + (y2 - y3)**2)
d31 = sqrt((x3 - x1)**2 + (y3 - y1)**2)
res = (d12 + d23 > d31) and (d12 + d31 > d23) and (d31 + d23 > d12)
if res:
  print("Forma um triangulo!")
else:
  print("Não é um triangulo!")
```

## 07. Classificação de triângulo pelos lados

Classificar um triângulo quanto aos lados, sendo que um triângulo com todos lados iguais é designado `Equilátero`, com todos os lados diferentes entre si é designado `Escaleno` e caso tenha apenas dois lados iguais entre si, designa-se `Isósceles`.

```python
# Resposta da questão 07
from math import sqrt
x1 = eval(input("x1: "))
y1 = eval(input("y1: "))
x2 = eval(input("x2: "))
y2 = eval(input("y2: "))
x3 = eval(input("x3: "))
y3 = eval(input("y3: "))
d12 = sqrt((x1 - x2)**2 + (y1 - y2)**2)
d23 = sqrt((x2 - x3)**2 + (y2 - y3)**2)
d31 = sqrt((x3 - x1)**2 + (y3 - y1)**2)
if (d12 == d23 == d31):
  print("Equilátero")
elif (d12 != d23 != d31):
  print("Escaleno")
else:
  print("Isóseles")
```

- Para transformar Celsius para Kelvin: `Tk = Tc + 273`
- Para transformar Celsius para Fahrenheit ou Fahrenheit para Celsius: `Tc = ((Tf - 32) / 9) * 5`
- Para transformar Kelvin para Celsius: `Tc = Tk - 273`

## 10. Ano bissexto

- A cada 4 anos o ano é bissexto, ou seja, todos os anos divisíveis por 4.
- A cada 100 anos o ano não é bissexto, apesar de ser divisível por 4.
- A cada 400 anos o ano é bissexto, apesar de ser divisível por 100.

```python
# Resposta da questão 10
ano = int(input("Digite um ano para saber se ele é bissexto: "))
cond1 = ano%4 == 0
cond2 = ano%100 == 0
cond3 = ano%400 == 0
if cond1 and not cond2:
  print(f"O ano {ano} é bissexto!")
elif cond2 and not cond3:
  print(f"O ano {ano} não é bissexto!")
elif cond3:
  print(f"O ano {ano} é bissexto!")
else:
  print(f"O ano {ano} não é bissexto!")
```

## 14. Existência e tipo de triângulo

Faça um programa que receba o tamanho dos três lados de um triângulo e diga se o triângulo existe e, se existir, qual o seu tipo.

- Um triângulo só pode existir quando todo lado é menor do que a soma dos outros dois lados.
- Um triângulo é dito equilátero se todos os lados são do mesmo tamanho.
- Um triângulo é dito isósceles se dois dos lados são do mesmo tamanho.
- Um triângulo é dito escaleno quando nenhum lado é do mesmo tamanho.

```python
# Resposta da questão 14
from math import sqrt
x1 = eval(input("x1: "))
y1 = eval(input("y1: "))
x2 = eval(input("x2: "))
y2 = eval(input("y2: "))
x3 = eval(input("x3: "))
y3 = eval(input("y3: "))
d12 = sqrt((x1 - x2)**2 + (y1 - y2)**2)
d23 = sqrt((x2 - x3)**2 + (y2 - y3)**2)
d31 = sqrt((x3 - x1)**2 + (y3 - y1)**2)
res = (d12 + d23 > d31) and (d12 + d31 > d23) and (d31 + d23 > d12)
if res:
  if (d12 == d23 == d31):
    print("Eexiste e é equilátero!")
  elif (d12 != d23 != d31):
    print("Existe e é escaleno!")
  else:
    print("Existe e é isóseles!")
else:
  print("Não é um triangulo!")
```

## 15. Sistema de equações com seis coeficientes

Considerando as fórmulas abaixo, leia o conjunto de coeficientes `(a, b, c, d, e, f)` e imprima a solução `x` e `y`. Antes de efetuar a divisão, verificar se ela pode ser feita. Em caso negativo, imprimir uma mensagem de que o sistema não tem solução.

$$x = \frac{ce - bf}{ae - bd}$$

$$y = \frac{af - cd}{ae - bd}$$

```python
# Resposta da questão 15
a = eval(input("a: "))
b = eval(input("b: "))
c = eval(input("c: "))
d = eval(input("d: "))
e = eval(input("e: "))
f = eval(input("f: "))
if (a*e - b*d) != 0:
  x = (c*e - b*f)/(a*e - b*d)
  y = (a*f - c*d)/(a*e - b*d)
  print(f"x = {x} e y = {y}")
else:
  print("O sistema não tem solução.")
```

## 16. Classificação de caractere

Ler um caractere e imprimir se é: um dígito (`0-9`), letra maiúscula (`A-Z`), letra minúscula (`a-z`), espaço ou outro símbolo.

```python
# Resposta da questão 16
import string
digito = list(string.digits)
letrasm = list(string.ascii_lowercase)
letrasM = list(string.ascii_uppercase)
espacos = list(string.whitespace)
simbolos = list(string.punctuation)
caractere = input("digite um caractere: ")
if len(list(caractere)) != 1:
  print("Você não digitou um caractere.")
elif caractere in digito:
  print("É um digito")
elif caractere in letrasm:
  print("É uma letra minúscula")
elif caractere in letrasM:
  print("É uma letra miaúscula")
elif caractere in espacos:
  print("É um espaço")
elif caractere in simbolos:
  print("É um símbolo")
```
