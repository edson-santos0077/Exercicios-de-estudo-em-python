### Dica

Comece assumindo que o primeiro elemento da lista é o maior ou menor. Depois, percorra o restante da lista comparando os valores.



````python
# Faça a questão 9 aqui.
def meu_max(lista):
  maior = lista[0]
  for i in lista:
    if i > maior:
      maior = i
  return maior

def meu_min(lista):
  menor = lista[0]
  for i in lista:
    if i < menor:
      menor = i
  return menor


pontuacoes = [45, 89, 32, 100, 12]
pontuacoes = [-5, -10, -2, -20]



print(f"O maior número é {meu_max(pontuacoes)}")
print(f"O maior número é {meu_min(pontuacoes)}")
````

O método `.index()` procura por um elemento dentro de uma lista e informa em qual posição ele está.

Crie um programa que tenha uma lista `convidados` e um `nome_buscado`.

O programa deve percorrer a lista e:

- imprimir o índice do nome, se ele for encontrado;
- imprimir uma mensagem informando que o nome não está na lista, caso ele não seja encontrado.

Não use `.index()`.

### Exemplo 1

```python
convidados = ["Ana", "Bruno", "Carlos", "Diana"]
nome_buscado = "Carlos"
```

Saída esperada:

```text
O convidado Carlos está no índice 2.
```

### Exemplo 2

```python
convidados = ["Ana", "Bruno", "Carlos", "Diana"]
nome_buscado = "Weskley"
```

Saída esperada:

```text
O convidado Weskley não está na lista.
```



````python
# Faça a questão 10 aqui.

def meu_index(lista, elemento):
  res = 0
  posicao = 0
  try:
    for c in lista:
      res += 1
      if elemento == c:
        posicao = res
  except:
    pass
  return  None if (posicao == 0) else (posicao - 1)

convidados = ["Ana", "Bruno", "Carlos", "Diana"]
nomes_buscado = "Carlos"
nomes_buscado = "Weskley"
verificador = meu_index(convidados, nomes_buscado)

if verificador == None:
  print(f"O convidado {nomes_buscado} não está na lista.")
else:
  print(f"O convidado Carlos está no índice {verificador}.")
````

Crie uma função chamada `calcular_media(lista)` que recebe uma lista de números e retorna a média dos valores.

Não use `sum` nem `len` diretamente. Reaproveite a lógica das questões 7 e 8, ou crie funções auxiliares.

### Exemplo 1

### Exemplo 2




````python
# Faça a questão 15 aqui.
def len_nova(meus_itens):
  cont = 0
  for a in meus_itens:
    cont += 1
  return cont

def minhasum(numeros):
  res = 0
  for n in numeros:
    res += n
  return res

def calcular_media(lista):
  media = minhasum(lista)/len_nova(lista)
  return media

notas = [7.5, 8.0, 6.5, 9.0]
notas = [10, 10, 10, 10, 10]
print(f"A média do aluno é {calcular_media(notas)}")
````

## 16) Removendo valores repetidos

Crie uma função chamada `remove_repetidos(lista)` que recebe uma lista e retorna uma nova lista sem valores repetidos, preservando a ordem original.

Exemplo:

```python
remove_repetidos([3, 1, 3, 2, 1, 4])
```

Resultado esperado:

```python
[3, 1, 2, 4]
```



````python
# Faça a questão 16 aqui.
def remove_repetidos(lista):
  copia = [l for l in lista] #list compreension cria uma nova lista
  indice = 0
  apagados = 0
  """A lógica é que primeiro eu percorro os elementos
     da lista e para cada elemento eu verifico os que
     estão a frente, se iguais eu apago o elemento da
     cópia. Os indices no final servem para acompanhar
     o indice do elemento do loop."""
  for i in lista:
    indicecopia = indice + 1
    for n in lista[(indice + 1):]:

      if n == i:
        del copia[indicecopia - apagados]
        apagados += 1

      indicecopia += 1

    indice +=1
  return copia

remove_repetidos([3, 1, 3, 2, 1, 4])
````
