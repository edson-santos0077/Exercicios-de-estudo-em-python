## 10) Experiências
São realizados N experimentos.
Cada linha contém:
- quantidade
- tipo (C, R ou S)

- Total de cobaias

```text

15 S
```

### Saída

---

## 17) Série com fatoriais


## 18) Pesquisa sobre o filme Senhor dos Anéis
No dia da estreia do filme "Senhor dos Anéis", uma emissora de TV realizou uma pesquisa com 20 espectadores. Cada espectador informou sua idade e sua opinião sobre o filme: `3` para excelente, `2` para bom e `1` para regular. Crie um algoritmo que receba a idade e a opinião dos 20 espectadores e calcule:

- a média das idades das pessoas que responderam excelente;
- a quantidade de pessoas que responderam regular;
- a porcentagem de pessoas que responderam bom entre todos os espectadores.

````python
# Resposta da questão 18

# Inicialização das variáveis para os cálculos
soma_idades_excelente = 0
qtd_excelente = 0
qtd_regular = 0
qtd_bom = 0

# Total de espectadores definido pelo enunciado
TOTAL_ESPECTADORES = 20

# Laço para ler os dados dos 20 espectadores
for i in range(TOTAL_ESPECTADORES):
    # Lendo a idade e a opinião de cada um
    idade = int(input(f"Digite a idade do espectador {i+1}: "))
    opiniao = int(input(f"Digite a opinião do espectador {i+1} (3-Excelente, 2-Bom, 1-Regular): "))

    # Processando os dados de acordo com a opinião
    if opiniao == 3:
        soma_idades_excelente += idade
        qtd_excelente += 1
    elif opiniao == 2:
        qtd_bom += 1
    elif opiniao == 1:
        qtd_regular += 1

#Média das idades das pessoas que responderam excelente
if qtd_excelente > 0:
    media_idades_excelente = soma_idades_excelente / qtd_excelente
else:
    media_idades_excelente = 0  # Evita divisão por zero caso ninguém vote 3

#Porcentagem de pessoas que responderam bom entre todos os espectadores
porcentagem_bom = (qtd_bom / TOTAL_ESPECTADORES) * 100

#Resultados

if qtd_excelente > 0:
    print(f"Média das idades (Excelente): {media_idades_excelente:.1f} anos")
else:
    print("Média das idades (Excelente): Ninguém respondeu Excelente.")

print(f"Quantidade de pessoas que responderam Regular: {qtd_regular}")
print(f"Porcentagem de pessoas que responderam Bom: {porcentagem_bom:.1f}%")
````

## 19) Fatorial de vários valores
Escreva um algoritmo que leia um número `N`, indicando quantos valores serão lidos a seguir. Para cada número lido, mostre o valor lido e o fatorial desse valor.

````python
# Resposta da questão 19

# Lê a quantidade de números que serão processados
n = int(input("Digite a quantidade de valores que serão lidos (N): "))

# Loop que vai rodar N vezes
for i in range(n):
    valor = int(input(f"Digite o {i+1}º número: "))

    # Inicializa a variável para calcular o fatorial
    fatorial = 1

    # Loop para calcular o fatorial do 'valor' lido
    for j in range(1, valor + 1):
        fatorial *= j

    # Mostra o valor lido e o seu respectivo fatorial
    print(f"Valor lido: {valor} -> Fatorial: {fatorial}")
````

## 20) Soma harmônica
Escreva um algoritmo que leia um valor `n` inteiro e positivo e calcule a seguinte soma:

```text
S = 1 + 1/2 + 1/3 + 1/4 + ... + 1/n
```

O algoritmo deve escrever cada termo gerado e o valor final de `S`.

````python
# Resposta da questão 20

# Lê o valor de n inteiro e positivo
n = int(input("Digite um valor inteiro e positivo para n: "))

# Inicializa a variável da soma
s = 0.0

# Laço para gerar os termos de 1 até n
for i in range(1, n + 1):
    termo = 1 / i
    s += termo
    # Exibe o termo atual formatado com 4 casas decimais
    print(f"Termo {i}: 1/{i} = {termo:.4f}")

# Exibe o valor final acumulado de S
print(f"Valor final de S = {s:.4f}")
````

## 21) Somatório duplo

$$\sum_{i=1}^{a} \sum_{j=1}^{b} \left((a \cdot b) + (j + i)^2\right)$$

## 22) Expressão alternada com n ímpar
Considerando `n` um valor de entrada numérico, inteiro e ímpar, escreva um programa que calcule e imprima o resultado da expressão:

```text
R = 1/n - 2/(n-1) + 3/(n-2) - ... + n
```

### Exemplos
```text
n = 3  -> R = 5.33
n = 5  -> R = 8.7
n = 7  -> R = 10.52
n = 9  -> R = 12.87
```

````python
# Resposta da questão 22
n = eval(input(""))
r = 0.0
for i in list(range(1, n + 1)):
  if i%2 != 0:
    r += i/(n - i + 1)
  else:
    r -= i/(n - i + 1)
print(f"n = {n} -> R = {r + n}")
````

## 23) Decaimento de material radioativo
Um determinado material radioativo perde metade de sua massa a cada 50 segundos. Dada como entrada a massa do material em gramas, escreva um programa que calcule o tempo necessário para que a massa desse material passe a ser inferior a 1 grama. Escreva a resposta em horas, minutos e segundos.

````python
# Resposta da questão 23

#Lendo a massa inicial em gramas
massa = float(input("Digite a massa inicial do material (em gramas): "))

tempo_segundos = 0

#O laço continua enquanto a massa for igual ou superior a 1 grama
while massa >= 1.0:
    massa /= 2          # Perde metade da massa
    tempo_segundos += 50 # Passaram-se 50 segundos

#Conversão do tempo total (segundos) para Horas, Minutos e Segundos
horas = tempo_segundos // 3600
resto = tempo_segundos % 3600
minutos = resto // 60
segundos = resto % 60

#Exibição dos Resultados
print("\nResultado do Decaimento")
print(f"Massa final: {massa:.4f} gramas")
print(f"Tempo necessário: {horas}h {minutos}min {segundos}s")
````
