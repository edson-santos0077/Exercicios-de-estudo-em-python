## 1) Contagem de 1 até N

Leia um número inteiro positivo `N` e imprima todos os números de `1` até `N`, um por linha.

### Entrada
```text
5
```

### Saída
```text
1
2
3
4
5
```


````python
# Faça a questão 1 aqui.
cont1 = 0

#Validar número como inteiro positivo
def validar_int(cond):
  while True :
    try:
      n = int(input())
      if cond(n):
        continue
      else:
        break
    except ValueError:
      pass
  return n

#Imprimir os numeros de 1 a n
n = validar_int(lambda x : x < 0)
while cont1 != n:
  cont1 += 1
  print(cont1)
````

## 8) Média com sentinela

Leia várias notas informadas pelo usuário. O programa deve parar quando o usuário digitar `-1`.

Ao final, imprima a média das notas digitadas antes do `-1`.

Considere que pelo menos uma nota válida será digitada.

### Entrada
```text
8
7
10
-1
```

### Saída
```text
Média: 8.33
```


````python
# Faça a questão 8 aqui.
#Bibliotecas
import ast

#Funções
def testar(n):
  try:
    n = ast.literal_eval(input())
    n = float(n)
  except ValueError:
    return False
  if (type(n) == float):
    if ((n >= 0) and (n <= 10)) or (n == -1):
      return n
    else:
      return False


#Declaração de variáveis
soma = media = elementos = n = 0

#Coleta das notas com filtros
while True:
  n = testar(n)

  if n == -1:
    break
  elif(n != False):
    soma += n
    elementos += 1
  else:
    print("Dígite uma nota válida")

#Calculo e apresentação da média
media = round((soma/elementos), 2)
print(f"Média: {media}")


````

Leia um número inteiro positivo em base decimal e converta para binário usando divisões sucessivas por `2`.

```

```

### Dica
A cada divisão por `2`, guarde o resto. Os restos aparecem na ordem inversa do número binário final.


````python
# Faça a questão 21 aqui.
n = validar_int(lambda x : x <= 0)
binario_invertido = []
indice = 0
binario = ''

#conversão no binário invertido
while n >= 1:
  binario_invertido += str(n % 2)
  n = n // 2

#invertendo o binário invertido
while indice != -(len(binario_invertido)):
  indice -= 1
  binario += binario_invertido[indice]

print(int(binario))
````

## 24) Raiz quadrada por aproximação

Leia um número positivo `N` e calcule uma aproximação para sua raiz quadrada usando o método de Newton.

Comece com:

```python
aproximacao = N / 2
```

Depois repita:

```python
nova = (aproximacao + N / aproximacao) / 2
```

A repetição deve parar quando a diferença entre `nova` e `aproximacao` for menor que `0.0001`.

### Entrada
```text
25
```

### Saída esperada aproximada
```text
Raiz aproximada: 5.0000
```


````python
# Faça a questão 24 aqui.

n = validar_int(lambda x : x < 0)


aproximacao = n / 2


while True:
    nova = (aproximacao + n / aproximacao) / 2
    diferenca = nova - aproximacao
    if diferenca < 0:
        diferenca = -diferenca
    if diferenca < 0.0001:
        aproximacao = nova
        break
    aproximacao = nova

print(f"Raiz aproximada: {aproximacao:.4f}")
````

## 25) Caixa eletrônico com quantidade limitada de cédulas

Um caixa eletrônico possui uma quantidade limitada de cédulas:

- 3 cédulas de 100;
- 2 cédulas de 50;
- 5 cédulas de 20;
- 10 cédulas de 10.

Leia um valor inteiro e tente montar esse valor usando as cédulas disponíveis, sempre priorizando as maiores cédulas.

Se for possível, imprima quantas cédulas de cada tipo foram usadas. Se não for possível formar exatamente o valor, imprima `Valor indisponível`.

### Entrada
```text
370
```

### Saída
```text
100: 3
50: 1
20: 1
10: 0
```

### Entrada
```text
385
```

### Saída
```text
Valor indisponível
```


````python
# Faça a questão 25 aqui.
valor = validar_int(lambda x: x < 0)
valor_restante = valor

# Definição da quantidade disponível de cada cédula
disponivel_100 = 3
disponivel_50 = 2
disponivel_20 = 5
disponivel_10 = 10

# Variáveis para armazenar a quantidade de cédulas utilizadas no saque
usadas_100 = 0
usadas_50 = 0
usadas_20 = 0
usadas_10 = 0

# --- Processamento das cédulas de 100 ---
necessarias_100 = valor_restante // 100
if necessarias_100 > disponivel_100:
    usadas_100 = disponivel_100
else:
    usadas_100 = necessarias_100
# Atualiza o valor restante que ainda precisa ser sacado
valor_restante = valor_restante - (usadas_100 * 100)

# --- Processamento das cédulas de 50 ---
necessarias_50 = valor_restante // 50
if necessarias_50 > disponivel_50:
    usadas_50 = disponivel_50
else:
    usadas_50 = necessarias_50
valor_restante = valor_restante - (usadas_50 * 50)

# --- Processamento das cédulas de 20 ---
necessarias_20 = valor_restante // 20
if necessarias_20 > disponivel_20:
    usadas_20 = disponivel_20
else:
    usadas_20 = necessarias_20
valor_restante = valor_restante - (usadas_20 * 20)

# --- Processamento das cédulas de 10 ---
necessarias_10 = valor_restante // 10
if necessarias_10 > disponivel_10:
    usadas_10 = disponivel_10
else:
    usadas_10 = necessarias_10
valor_restante = valor_restante - (usadas_10 * 10)

# --- Verificação e Exibição do Resultado ---
# Se o valor restante zerou, significa que foi possível formar o valor exato
if valor_restante == 0:
    print(f"100: {usadas_100}")
    print(f"50: {usadas_50}")
    print(f"20: {usadas_20}")
    print(f"10: {usadas_10}")
else:
    print("Valor indisponível")
````
