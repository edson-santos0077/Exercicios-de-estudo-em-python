## 1. Implementação de expressão aritmética

Linearize a expressão a seguir e implemente seu cálculo em Python. O programa deve ler o valor de `x` e calcular `Z`.

$$Z = \frac{2x^2 - 5x + 8}{3x + 1} + \frac{\sqrt{x + 4}}{2}$$

**Exemplo:**

Entrada: `x = 5`

Saída esperada aproximada: `Z = 4.949`

```python
# Resposta da questão 1
from math import sqrt
x = eval(input("x: "))
a = 2*x**2 - 5*x + 8
b = 3*x + 1
c = sqrt(x + 4)/2
z = a/b + c
print(f"z = {z}")
```

## 9. Expressao com potências e raiz quadrada

Linearize a expressão e implemente em Python. O programa deve ler os valores de `a`, `b` e `c` e calcular `R`.

$$R = \frac{a^2 + 2ab + b^2}{\sqrt{c + 10}} - \frac{a - b}{c}$$

**Exemplo:**

Entrada: `a = 4`, `b = 2`, `c = 6`

Saída esperada aproximada: `R = 8.667`

```python
# Resposta da questão 9
from math import sqrt
a = eval(input("a: "))
b = eval(input("b: "))
c = eval(input("c: "))
x = (a**2 + 2*a*b + b**2)
y = sqrt(c + 10)
z = (a - b)/c
r = x/y - z
print(f"R = {r}")
```

## 14. Erro quadratico médio - MSE

Uma medida de perda comum em Inteligência Artificial é o erro quadratico médio, ou MSE. Escreva um programa que leia três valores reais esperados (`y1`, `y2`, `y3`) e três valores previstos (`p1`, `p2`, `p3`). Calcule o MSE.

$$MSE = \frac{(y_1 - p_1)^2 + (y_2 - p_2)^2 + (y_3 - p_3)^2}{3}$$

**Exemplo:**

Entrada: `y1 = 3`, `y2 = 5`, `y3 = 2`, `p1 = 2.5`, `p2 = 5.5`, `p3 = 1`

Saída esperada: `MSE = 0.5`

```python
# Resposta da questão 14
y1 = eval(input("y1: "))
y2 = eval(input("y2: "))
y3 = eval(input("y3: "))
p1 = eval(input("p1: "))
p2 = eval(input("p2: "))
p3 = eval(input("p3: "))
mse = ((y1 - p1)**2 + (y2 - p2)**2 + (y3 - p3)**2)/3
print(f"MSE = {mse}")
```

## 15. Erro absoluto médio - MAE

Outra medida de erro usada em modelos de previsão é o erro absoluto médio, ou MAE. Leia três valores esperados e três valores previstos. Depois calcule o MAE.

$$MAE = \frac{|y_1 - p_1| + |y_2 - p_2| + |y_3 - p_3|}{3}$$

**Exemplo:**

Entrada: `y1 = 10`, `y2 = 20`, `y3 = 30`, `p1 = 12`, `p2 = 18`, `p3 = 33`

Saída esperada aproximada: `MAE = 2.333`

Para pegar o valor absoluto `|x|` use `math.abs(x)`

```python
# Resposta da questão 15
from math import fabs
y1 = eval(input("y1: "))
y2 = eval(input("y2: "))
y3 = eval(input("y3: "))
p1 = eval(input("p1: "))
p2 = eval(input("p2: "))
p3 = eval(input("p3: "))
mae = (fabs(y1-p1) + fabs(y2-p2) + fabs(y3-p3))/3
print(f"MAE = {mae}")
```

## 16. Normalização min-max

Em pré-processamento de dados, uma forma simples de normalizar um valor `x` é usar o menor valor `minimo` e o maior valor `maximo` do conjunto. Leia `x`, `minimo` e `maximo`, e calcule o valor normalizado.

$$x_{norm} = \frac{x - minimo}{maximo - minimo}$$

**Exemplo:**

Entrada: `x = 40`, `minimo = 20`, `maximo = 100`

Saída esperada: `x_norm = 0.25`

```python
# Resposta da questão 16
x = eval(input("x: "))
minimo = eval(input("minimo: "))
maximo = eval(input("maximo: "))
x_norm = (x - minimo)/(maximo - minimo)
print(f"X_norm = {x_norm}")
```

Escreva um programa que leia o preço original de um produto, o percentual de desconto e o percentual de imposto. Primeiro aplique o desconto e depois aplique o imposto sobre o valor com desconto.

$$preco\_desconto = preco(1 - desconto)$$

$$preco\_final = preco\_desconto(1 + imposto)$$

**Exemplo:**
