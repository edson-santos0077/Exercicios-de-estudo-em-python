## Exercício 2

Crie uma função chamada `calcula_descontos`.

Ela deve receber uma lista de preços e retornar uma nova lista com desconto de 10% aplicado em cada preço.

Use `for` dentro da função.


```python
def calcula_descontos(precos):
    # escreva sua solução aqui
    descontados = []
    for n in precos:
      descontados.append(n * 0.9)
    return descontados


print(calcula_descontos([100, 50, 80]))
# esperado: [90.0, 45.0, 72.0]
```

## Exercício 3

Crie uma função chamada `classifica_notas`.

Ela deve receber uma lista de notas e retornar uma lista de textos:

- `Aprovado`, se a nota for maior ou igual a 7;
- `Recuperação`, se a nota for maior ou igual a 5 e menor que 7;
- `Reprovado`, se a nota for menor que 5.

Use `for` e `if/elif/else` dentro da função.


```python
def classifica_notas(notas):
    # escreva sua solução aqui
    lista = []
    for n in notas:
      if n > 7:
        lista.append("Aprovado")
      elif n >= 5:
        lista.append("Recuperação")
      else:
        lista.append("Reprovado")
    return lista


print(classifica_notas([8, 6, 4, 10]))
# esperado: ['Aprovado', 'Recuperação', 'Reprovado', 'Aprovado']
```
