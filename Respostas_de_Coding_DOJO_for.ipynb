## Problema 1 — Soma de 1 até n (nível: aquecimento)

Implemente `soma_ate(n)` que retorna a soma dos números inteiros de `1` até `n`, inclusive.

**Exemplo**

```python
soma_ate(5)  # 1 + 2 + 3 + 4 + 5 = 15
```

**Restrições e dicas**
- Use `for`.
- Não use a fórmula `n * (n + 1) / 2`.
- Considere `n >= 1`.


````python
def soma_ate(n):
    """Retorna a soma dos inteiros de 1 até n."""
    soma = 0 #Definindo variável
    for numero in range(1, n + 1):
      soma += numero #Soma número ao total em cada iteração

    return soma
````

````python
# Testes rápidos
try:
    assert soma_ate(1) == 1
    assert soma_ate(5) == 15
    assert soma_ate(10) == 55
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 2 — Quantidade de pares (nível: aquecimento)

Implemente `conta_pares(nums)` que recebe uma lista de números inteiros e retorna quantos deles são pares.

**Exemplo**

```python
conta_pares([2, 7, 4, 9, 10])  # retorna 3
```

**Restrições e dicas**
- Use `for` para percorrer a lista.
- Use o operador `%` para testar se um número é par.


````python
def conta_pares(nums):
    """Retorna a quantidade de números pares na lista."""
    cont = 0
    if nums:
      for i in range(len(nums)):
        if nums[i] % 2 == 0:
          cont += 1

    return cont
````

````python
# Testes rápidos
try:
    assert conta_pares([]) == 0
    assert conta_pares([1, 3, 5]) == 0
    assert conta_pares([2, 6, 4, 8, 10]) == 5
    assert conta_pares([0, -2, -3, 8]) == 3
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 3 — Maior valor sem usar `max` (nível: básico)

Implemente `maior_valor(nums)` que recebe uma lista não vazia de números e retorna o maior valor.

**Restrições e dicas**
- Use `for`.
- Não use `max`.
- Comece assumindo que o primeiro elemento é o maior e compare com os demais.
- A lista sempre terá pelo menos um elemento.


````python
def maior_valor(nums):
    """Retorna o maior valor de uma lista não vazia."""
    # escreva sua solução aqui
    mai = nums[0]
    for i in nums:
      if mai <= i:
        mai = i
    return mai

````

````python
# Testes rápidos
try:
    assert maior_valor([5]) == 5
    assert maior_valor([1, 8, 3, 2]) == 8
    assert maior_valor([-10, -3, -7]) == -3
    assert maior_valor([4, 4, 4]) == 4
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 4 — Média dos positivos (nível: básico/médio)

Implemente `media_positivos(nums)` que retorna a média apenas dos números positivos da lista.

Se não houver nenhum número positivo, retorne `0`.

**Exemplo**

```python
media_positivos([-2, 4, 6, -1])  # retorna 5.0
```

**Restrições e dicas**
- Use `for`.
- Mantenha um **acumulador** para a soma e um **contador** para a quantidade.
- Cuidado com divisão por zero.


````python
def media_positivos(nums):
    """Retorna a média dos números positivos ou 0 se não houver positivos."""
    # escreva sua solução aqui
    media = 0
    cont = 0

    for i in nums:
      if i > 0:
        media += i
        cont += 1

    if cont == 0:
      return 0
    else:
      media = media/cont
      return media





````

````python
# Testes rápidos
try:
    assert media_positivos([-2, 4, 6, -1]) == 5.0
    assert media_positivos([1, 2, 3]) == 2.0
    assert media_positivos([-5, 0, -1]) == 0
    assert media_positivos([10]) == 10.0
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 5 — Sequência acumulada (nível: médio)

Implemente `somas_acumuladas(nums)` que recebe uma lista de números e retorna uma nova lista com as somas acumuladas.

**Exemplo**

```python
somas_acumuladas([2, 4, 1, 3])  # retorna [2, 6, 7, 10]
```

**Restrições e dicas**
- Use `for`.
- Mantenha uma variável com a soma acumulada até o momento.
- A lista original não deve ser alterada.


````python
def somas_acumuladas(nums):
    """Retorna uma lista com as somas acumuladas."""
    # escreva sua solução aqu
    nl = []
    acumulo = 0
    for i in nums:
      acumulo += i
      nl.append(acumulo)
    return nl
````

````python
# Testes rápidos
try:
    assert somas_acumuladas([]) == []
    assert somas_acumuladas([2, 4, 1, 3]) == [2, 6, 7, 10]
    assert somas_acumuladas([-1, 5, -2]) == [-1, 4, 2]
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 6 — Compactação RLE (Run-Length Encoding) (nível: médio)

Implemente `rle_encode(s)` que **compacta** uma string substituindo sequências de caracteres repetidos por `<caractere><contagem>`.
Ex.: `AAABCC` → `A3B1C2`.

**Regras**
- Trabalhe **caractere a caractere** com `for`.
- Não use `itertools.groupby`.
- Sempre inclua a contagem, mesmo quando ela for `1`.
- A string vazia deve retornar `""`.


````python
def rle_encode(s):
    """Retorna a codificação RLE sempre incluindo a contagem."""
    # escreva sua solução aqui
    resposta = ""

    if s:
      palavra = s[0]
      cont = 1

      for i in s[1:]:
        if i == palavra:
            cont += 1
        else:
            resposta += palavra + str(cont)
            palavra = i
            cont = 1

      resposta += palavra + str(cont)

    return resposta

````

````python
# Testes rápidos
try:
    assert rle_encode("") == ""
    assert rle_encode("A") == "A1"
    assert rle_encode("AAABCC") == "A3B1C2"
    assert rle_encode("ABBBCCCCDAA") == "A1B3C4D1A2"
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````

## Problema 7 — Soma em matriz com `for` encadeado (nível: médio/alto)

Implemente `soma_diagonal_secundaria(matriz)` que recebe uma matriz quadrada e retorna a soma dos elementos da diagonal secundária.

Na matriz abaixo, a diagonal secundária é formada por `3`, `5` e `7`:

```python
[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

**Restrições e dicas**
- Use `for`.
- A matriz sempre será quadrada.
- Pense na relação entre a linha `i` e a coluna da diagonal secundária.


````python
def soma_diagonal_secundaria(matriz):
    """Retorna a soma da diagonal secundária de uma matriz quadrada."""
    #Definindo variáveis
    soma = 0
    coluna = -1 #começamos com a última coluna

    #Para cada linha é somado elemento da coluna determinada
    for linha in matriz:
      soma += linha[coluna]
      coluna -= 1 #para cada iteração avançamos uma coluna da direita para a esqerda

    return soma
````

````python
# Testes rápidos
try:
    assert soma_diagonal_secundaria([[1]]) == 1
    assert soma_diagonal_secundaria([[1, 2], [3, 4]]) == 5  # 2 + 3
    assert soma_diagonal_secundaria([[1,2,3], [4,5,6], [7,8,9]]) == 15
    assert soma_diagonal_secundaria([[10,2,3,4], [5,20,7,8], [9,10,30,12], [13,14,15,40]]) == 34
    print("Tudo certo!")
except AssertionError:
    print("Ainda há algo para ajustar.")
````
