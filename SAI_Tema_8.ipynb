Apresentação sobre funções chamando funções como argumento e funções lambda.

```python
#Variáveis
lista = [2, 3, 4, -1]

def dobro(lista):
  #retorna a lista de resposta gerada por uma compressão de lista
  return [l*2 for l in lista]

def eh_positivo(lista):
  #retorna a lista de resposta gerada por uma compressão de lista
  return [l > 0 for l in lista]

#Corpo do código
print(dobro(lista))
print(eh_positivo(lista))
```

```python
#Variáveis
lista = [2, 3, 4, -1]

#Funções lambda
dobro = lambda x : x*2
eh_positivo = lambda x : (x > 0)


#Corpo do código
print(dobro(lista))
print(eh_positivo(lista))
```

```python

#Outros exemplos de funções lambda
k = 0

#valor padrão
dobro = lambda x = 1 : x*2
print(dobro())




#busca no escopo LEGB
eh_positivo = lambda x : x > k #isso é um L
print(eh_positivo(2))
print(eh_positivo)
```

```python
#Funções recebendo funções como argumento(callback)
def aplica_operacao(lista, funcao):
  #retorna a lista de resposta gerada por uma compressão de lista
  return [funcao(l) for l in lista]

#Exemplo simples
def divisao_por_dois(x):
  return x/2

def soma(callback, x, y):
  return callback(x) + y

print(soma(divisao_por_dois, 2, 3))
```

```python
#Variáveis
li = [2, 3, 4, -1]

#Funções recebendo funções como argumento(callback)
def aplica_operacao(lista, funcao):
  #retorna a lista de resposta gerada por uma compressão de lista
  return [funcao(l) for l in lista]

#Funções lambda
dobro = lambda x : x*2
eh_positivo = lambda x : (x > 0)


#Corpo do código
print(aplica_operacao(li, dobro))
print(aplica_operacao(li, eh_positivo))
```
